```python

import numpy as np

# Step 1: Dataset (XOR)
X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])

y = np.array([
    [0],
    [1],
    [1],
    [0]
])

# Step 2: Initialize weights and biases
np.random.seed(42)

W1 = np.random.randn(2, 3)
b1 = np.zeros((1, 3))

W2 = np.random.randn(3, 1)
b2 = np.zeros((1, 1))

# Step 3: Sigmoid activation function
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

# Step 4: Training parameters
learning_rate = 0.5
epochs = 10000

# Step 5: Backpropagation training
for epoch in range(epochs):

    # Forward propagation
    hidden_input = np.dot(X, W1) + b1
    hidden_output = sigmoid(hidden_input)

    final_input = np.dot(hidden_output, W2) + b2
    predicted_output = sigmoid(final_input)

    # Calculate error
    error = predicted_output - y

    # Backward propagation
    d_output = error * predicted_output * (1 - predicted_output)

    dW2 = np.dot(hidden_output.T, d_output) / len(X)
    db2 = np.mean(d_output, axis=0, keepdims=True)

    d_hidden = np.dot(d_output, W2.T)
    d_hidden *= hidden_output * (1 - hidden_output)

    dW1 = np.dot(X.T, d_hidden) / len(X)
    db1 = np.mean(d_hidden, axis=0, keepdims=True)

    # Update weights and biases
    W2 -= learning_rate * dW2
    b2 -= learning_rate * db2

    W1 -= learning_rate * dW1
    b1 -= learning_rate * db1

# Step 6: Test the trained network
hidden_output = sigmoid(np.dot(X, W1) + b1)
predictions = sigmoid(np.dot(hidden_output, W2) + b2)

print("ANN using Backpropagation")
print("-------------------------")
print("Input\tExpected\tPredicted")

for i in range(len(X)):
    result = 1 if predictions[i][0] >= 0.5 else 0

    print(X[i], "\t", y[i][0], "\t\t", result)

print("\nTraining Completed Successfully!")
```
